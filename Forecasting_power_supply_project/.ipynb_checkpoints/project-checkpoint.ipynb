```python
df=df.sort_values('Datetime')
```

```python
#Overall Energy Consumption Trend
plt.figure(figsize=(40,20))
plt.plot(df['Datetime'],df['PJMW_MW'])
plt.xlabel('Date')
plt.ylabel('PJMW_MW')
plt.title("energy consumptiom per hour")
plt.show()
```

```python
#Daily Average Consumption
```

```python
daily_avg = (
    df.set_index('Datetime')['PJMW_MW']
    .resample('D')
    .mean()
)
```

```python
daily_avg
```

```python
plt.figure(figsize=(10,5))
plt.plot(daily_avg.index,daily_avg.values)
plt.xlabel('Date')
plt.ylabel('energy consumption')
plt.title('energy consumption daily')
```

```python
monthly_avg = (
    df.set_index('Datetime')['PJMW_MW']
    .resample('ME')
    .mean()
)
```

```python
plt.figure(figsize=(16, 6))

plt.plot(monthly_avg.index, monthly_avg.values)

plt.xlabel('Date')
plt.ylabel('Average Energy Consumption (MW)')
plt.title('Monthly Average Energy Consumption')

plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.histplot(
    df['PJMW_MW'],
    bins=50,
    kde=True
)

plt.xlabel('Energy Consumption (MW)')
plt.ylabel('Frequency')
plt.title('Distribution of Energy Consumption')

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(x=df['PJMW_MW'])

plt.xlabel('Energy Consumption (MW)')
plt.title('Boxplot of Energy Consumption')

plt.show()
```

```python
df['hour'] = df['Datetime'].dt.hour
df['day'] = df['Datetime'].dt.day
df['day_of_week'] = df['Datetime'].dt.dayofweek
df['month'] = df['Datetime'].dt.month
df['year'] = df['Datetime'].dt.year
```

```python
df.head()
```

```python
hourly_avg = (
    df.groupby('hour')['PJMW_MW']
    .mean()
)
```

```python
hourly_avg
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    hourly_avg.index,
    hourly_avg.values,
    marker='o'
)

plt.xlabel('Hour of Day')
plt.ylabel('Average Energy Consumption (MW)')
plt.title('Average Energy Consumption by Hour')

plt.xticks(range(24))

plt.show()
```

```python
day_names = {
    0: 'Monday',
    1: 'Tuesday',
    2: 'Wednesday',
    3: 'Thursday',
    4: 'Friday',
    5: 'Saturday',
    6: 'Sunday'
}
```

```python
weekday_avg = (
    df.groupby('day_of_week')['PJMW_MW']
    .mean()
)
```

```python
weekday_avg 
```

```python
plt.figure(figsize=(10, 5))

plt.bar(
    [day_names[i] for i in weekday_avg.index],
    weekday_avg.values
)

plt.xlabel('Day of Week')
plt.ylabel('Average Energy Consumption (MW)')
plt.title('Average Energy Consumption by Day of Week')

plt.xticks(rotation=30)

plt.show()
```

```python
month_avg = (
    df.groupby('month')['PJMW_MW']
    .mean()
)
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    month_avg.index,
    month_avg.values,
    marker='o'
)

plt.xlabel('Month')
plt.ylabel('Average Energy Consumption (MW)')
plt.title('Average Energy Consumption by Month')

plt.xticks(
    range(1, 13),
    ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
     'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
)

plt.show()
```

```python
hour_month = df.pivot_table(
    values='PJMW_MW',
    index='hour',
    columns='month',
    aggfunc='mean'
)
```

```python
hour_month
```

```python
plt.figure(figsize=(14, 8))

sns.heatmap(
    hour_month,
    cmap='YlOrRd'
)

plt.xlabel('Month')
plt.ylabel('Hour of Day')
plt.title('Average Energy Consumption by Hour and Month')

plt.show()
```

```python
weekday_hour = df.pivot_table(
    values='PJMW_MW',
    index='day_of_week',
    columns='hour',
    aggfunc='mean'
)
```

```python
weekday_hour
```

```python
plt.figure(figsize=(14, 6))

sns.heatmap(
    weekday_hour,
    cmap='YlOrRd'
)

plt.xlabel('Hour of Day')
plt.ylabel('Day of Week')
plt.title('Average Energy Consumption by Day and Hour')

plt.yticks(
    range(7),
    ['Monday', 'Tuesday', 'Wednesday',
     'Thursday', 'Friday', 'Saturday', 'Sunday'],
    rotation=0
)

plt.show()
```

```python
from pandas.tseries.holiday import USFederalHolidayCalendar

calendar = USFederalHolidayCalendar()

holidays = calendar.holidays(
    start=df['Datetime'].min(),
    end=df['Datetime'].max()
)

df['is_holiday'] = (
    df['Datetime'].dt.normalize().isin(holidays)
).astype(int)
```

```python
df['is_holiday'].value_counts()
```

```python
holiday_avg = (
    df.groupby('is_holiday')['PJMW_MW']
    .mean()
)

holiday_avg
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    ['Non-Holiday', 'Holiday'],
    holiday_avg.values
)

plt.ylabel('Average Energy Consumption (MW)')
plt.title('Holiday vs Non-Holiday Consumption')

plt.show()
```

```python
yearly_avg = (
    df.groupby('year')['PJMW_MW']
    .mean()
)
```

```python
plt.figure(figsize=(14, 5))

plt.plot(
    yearly_avg.index,
    yearly_avg.values,
    marker='o'
)

plt.xlabel('Year')
plt.ylabel('Average Energy Consumption (MW)')
plt.title('Average Energy Consumption by Year')

plt.show()
```
