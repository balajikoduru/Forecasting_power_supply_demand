Outlier analysis

```python
Q1 = df['PJMW_MW'].quantile(0.25)
Q3 = df['PJMW_MW'].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df['PJMW_MW'] < lower_limit) | (df['PJMW_MW'] > upper_limit)]

print('Q1:',Q1)
print('Q3:',Q3)
print('IQR:',IQR)
print('Lower limit:',lower_limit)
print('Upper limit:',upper_limit)
print('Number of Outliers:',len(outliers))
```

Time-Series Analysis

```python
plt.figure(figsize=(15, 5))
plt.plot(df['Datetime'], df['PJMW_MW'])
plt.title('Power Demand Over Time')
plt.xlabel('Datetime')
plt.ylabel('Power Demand (MW)')
plt.show()
```

```python
yearly_demand = df.groupby(df['Datetime'].dt.year)['PJMW_MW'].mean()
yearly_demand
```

```python
plt.figure(figsize=(12, 5))

plt.plot(yearly_demand.index, yearly_demand.values, marker='o')

plt.title('Average Power Demand by Year')
plt.xlabel('Year')
plt.ylabel('Average Power Demand (MW)')

plt.show()
```

```python
monthly_demand = df.groupby(df['Datetime'].dt.month)['PJMW_MW'].mean()
monthly_demand
```

```python
plt.figure(figsize=(10, 5))

plt.plot(monthly_demand.index, monthly_demand.values, marker='o')

plt.title('Average Power Demand by Month')
plt.xlabel('Month')
plt.ylabel('Average Power Demand (MW)')
plt.xticks(range(1, 13))

plt.show()
```

```python
daily_demand = df.groupby(df['Datetime'].dt.dayofweek)['PJMW_MW'].mean()
daily_demand
```

```python
plt.figure(figsize=(10, 5))

plt.plot(daily_demand.index, daily_demand.values, marker='o')

plt.title('Average Power Demand by Day of Week')
plt.xlabel('Day of Week')
plt.ylabel('Average Power Demand (MW)')

plt.xticks(range(7),
           ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
            'Friday', 'Saturday', 'Sunday'])

plt.show()
```

```python
hourly_demand = df.groupby(df['Datetime'].dt.hour)['PJMW_MW'].mean()
hourly_demand
```

```python
plt.figure(figsize=(10, 5))

plt.plot(hourly_demand.index, hourly_demand.values, marker='o')

plt.title('Average Power Demand by Hour')
plt.xlabel('Hour of Day')
plt.ylabel('Average Power Demand (MW)')

plt.xticks(range(24))

plt.show()
```

Correlation / Relationship Analysis

```python
df['Year'] = df['Datetime'].dt.year
df['Month'] = df['Datetime'].dt.month
df['Day'] = df['Datetime'].dt.day
df['Hour'] = df['Datetime'].dt.hour
df['DayOfWeek'] = df['Datetime'].dt.dayofweek
df['IsWeekend'] = df['DayOfWeek'].isin([5, 6]).astype(int)
```

```python
df.head()
```

```python
corr = df[['Year', 'Month', 'Day', 'Hour', 'DayOfWeek', 'IsWeekend', 'PJMW_MW']].corr()

plt.figure(figsize=(10, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
df = df.sort_values('Datetime').reset_index(drop=True)
```

```python
df[['Datetime', 'PJMW_MW']].head()
```

Lag features

```python
df['Lag_1'] = df['PJMW_MW'].shift(1)
df['Lag_24'] = df['PJMW_MW'].shift(24)
df['Lag_168'] = df['PJMW_MW'].shift(168)
```

```python
df[['Datetime', 'PJMW_MW', 'Lag_1', 'Lag_24', 'Lag_168']].head(170)
```

Remove missing values from lag creation

```python
df = df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

Train-test split

```python
train = df[df['Datetime'].dt.year < 2018]
test = df[df['Datetime'].dt.year >= 2018]
```

```python
print("Train shape:", train.shape)
print("Test shape:", test.shape)
```

```python
print("Train period:", train['Datetime'].min(), "to", train['Datetime'].max())
print("Test period:", test['Datetime'].min(), "to", test['Datetime'].max())
```

Separate X and y

```python
features = ['Year', 'Month', 'Day', 'Hour', 'DayOfWeek',
            'IsWeekend', 'Lag_1', 'Lag_24', 'Lag_168']

X_train = train[features]
y_train = train['PJMW_MW']

X_test = test[features]
y_test = test['PJMW_MW']
```

```python
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```
