```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Pandas is used for loading and manipulating the dataset. NumPy is used for numerical operations. Matplotlib and Seaborn are used to create graphs and visualize the patterns in electricity consumption.

```python
data = pd.read_csv('power_supply.csv')

data.head()
```

The PJM hourly energy consumption dataset is loaded into a Pandas DataFrame using the read_csv() function. The head() function displays the first five records to understand the structure of the dataset.

```python
data.shape
```

The shape function gives the number of rows and columns in the dataset. This helps us understand the size of the dataset.

```python
data.columns
```

This displays the names of all columns present in the dataset. The dataset contains Datetime and PJMW_MW.

```python
data.dtypes
```

This checks the data type of each column. The Datetime column should be in datetime format, while PJMW_MW contains numerical power consumption values.

```python
data.info()
```

The info() function provides information about the dataset, including the number of entries, column names, data types and non-null values. It helps us identify missing values and understand the structure of the dataset.

```python
data.describe()
```

The describe() function provides statistical information about the numerical column. It shows the count, mean, standard deviation, minimum, maximum and quartile values of power consumption.
This helps us understand the general distribution of electricity demand.

```python
data.isnull().sum()
```

This checks whether any values are missing in each column. Missing values can affect analysis and forecasting, so they need to be identified before further analysis.

```python
data.duplicated().sum()
```

This checks whether duplicate records are present in the dataset. Duplicate observations can affect the accuracy of analysis and therefore need to be identified.

```python
data['Datetime'] = pd.to_datetime(data['Datetime'])

data.dtypes
```

The Datetime column is converted into Pandas datetime format. This is important because time-based analysis requires extracting information such as hour, day, month and year from the date and time.

```python
print("Start Date:", data['Datetime'].min())
print("End Date:", data['Datetime'].max())
```

This identifies the beginning and ending dates available in the dataset. Understanding the time range is important for a time-series forecasting project.

```python
data = data.sort_values('Datetime')

data.head()
```

The dataset is sorted according to the Datetime column so that the observations are arranged chronologically. Correct chronological order is important for time-series analysis and forecasting.

```python
data['PJMW_MW'].isnull().sum()
```

This specifically checks whether the power consumption column contains missing values. Missing consumption values need to be handled before building a forecasting model.

```python
plt.figure(figsize=(15,5))

plt.plot(data['Datetime'], data['PJMW_MW'])

plt.xlabel('Date')
plt.ylabel('Power Consumption (MW)')
plt.title('Hourly Power Consumption Over Time')

plt.show()
```

This line plot shows the variation in electricity consumption over the entire period. It helps us identify overall trends, fluctuations and seasonal patterns in power demand.

```python
data['Year'] = data['Datetime'].dt.year
data['Month'] = data['Datetime'].dt.month
data['Day'] = data['Datetime'].dt.day
data['Hour'] = data['Datetime'].dt.hour
data['DayOfWeek'] = data['Datetime'].dt.dayofweek

data.head()
```

New time-based columns are created from the Datetime column.
These features help us study how electricity demand changes according to different time periods.

```python
data['DayName'] = data['Datetime'].dt.day_name()

data[['Datetime', 'DayName']].head()
```

The day name is extracted from the datetime column. This allows us to compare electricity consumption between Monday, Tuesday, Wednesday and other days of the week.

```python
hourly_consumption = data.groupby('Hour')['PJMW_MW'].mean()

hourly_consumption
```

The data is grouped according to the hour of the day and the average power consumption is calculated for each hour. This helps us understand which hours generally have higher or lower electricity demand.

```python
plt.figure(figsize=(10,5))

plt.plot(hourly_consumption.index, hourly_consumption.values, marker='o')

plt.xlabel('Hour of Day')
plt.ylabel('Average Power Consumption (MW)')
plt.title('Average Power Consumption by Hour')

plt.xticks(range(24))

plt.show()
```

This graph shows the average electricity consumption for each hour of the day. It helps identify peak demand hours and low-demand hours.

```python
daily_consumption = data.groupby('DayName')['PJMW_MW'].mean()

daily_consumption
```

The dataset is grouped by day of the week and the average electricity consumption is calculated. This helps us understand whether electricity demand differs between weekdays and weekends.

```python
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday',
        'Friday', 'Saturday', 'Sunday']

daily_consumption = data.groupby('DayName')['PJMW_MW'].mean()

daily_consumption = daily_consumption.reindex(days)

daily_consumption
```

The days are arranged in their natural weekly order from Monday to Sunday. This makes the analysis and visualization easier to interpret.

```python
plt.figure(figsize=(10,5))

plt.bar(daily_consumption.index, daily_consumption.values)

plt.xlabel('Day of Week')
plt.ylabel('Average Power Consumption (MW)')
plt.title('Average Power Consumption by Day of Week')

plt.xticks(rotation=45)

plt.show()
```

This bar graph compares average electricity consumption across different days of the week. It helps us identify differences between weekdays and weekends.

```python
data['DayType'] = data['DayOfWeek'].apply(
    lambda x: 'Weekend' if x >= 5 else 'Weekday'
)

data.groupby('DayType')['PJMW_MW'].mean()
```

The days are divided into two groups: weekdays and weekends. The average electricity consumption for both groups is calculated to determine whether demand changes during weekends.

```python
day_type_consumption = data.groupby('DayType')['PJMW_MW'].mean()

plt.figure(figsize=(7,5))

plt.bar(day_type_consumption.index, day_type_consumption.values)

plt.xlabel('Day Type')
plt.ylabel('Average Power Consumption (MW)')
plt.title('Weekday vs Weekend Power Consumption')

plt.show()
```

This graph compares average power consumption between weekdays and weekends. It provides a simple visual understanding of weekly demand differences.

```python
monthly_consumption = data.groupby('Month')['PJMW_MW'].mean()

monthly_consumption
```

The dataset is grouped by month and the average electricity consumption is calculated. This helps identify monthly and seasonal variations in power demand.

```python
plt.figure(figsize=(10,5))

plt.plot(monthly_consumption.index, monthly_consumption.values, marker='o')

plt.xlabel('Month')
plt.ylabel('Average Power Consumption (MW)')
plt.title('Average Power Consumption by Month')

plt.xticks(range(1,13))

plt.show()
```

This graph shows the average electricity consumption for each month. It helps identify seasonal changes in electricity demand.

```python
yearly_consumption = data.groupby('Year')['PJMW_MW'].mean()

yearly_consumption
```

The data is grouped by year and the average electricity consumption is calculated. This helps us understand long-term changes in electricity demand.

```python
plt.figure(figsize=(10,5))

plt.plot(yearly_consumption.index, yearly_consumption.values, marker='o')

plt.xlabel('Year')
plt.ylabel('Average Power Consumption (MW)')
plt.title('Yearly Average Power Consumption')

plt.show()
```

This graph shows the yearly average electricity consumption and helps identify long-term increases or decreases in power demand.

```python
monthly_yearly = data.groupby(['Year', 'Month'])['PJMW_MW'].mean().reset_index()

monthly_yearly.head()
```

The data is grouped by both year and month. This allows us to compare monthly electricity demand across different years and understand how seasonal patterns change over time.

```python
plt.figure(figsize=(12,6))

sns.lineplot(
    data=monthly_yearly,
    x='Month',
    y='PJMW_MW',
    hue='Year'
)

plt.xlabel('Month')
plt.ylabel('Average Power Consumption (MW)')
plt.title('Monthly Power Consumption Pattern Across Years')

plt.show()
```

This graph compares monthly electricity consumption patterns across different years. It helps us understand whether seasonal patterns remain similar or change over time.

```python
hour_month = data.groupby(['Month', 'Hour'])['PJMW_MW'].mean().reset_index()

hour_month.head()
```

The data is grouped by both month and hour. This allows us to study how the daily electricity demand pattern changes across different months.

```python
heatmap_data = data.pivot_table(
    values='PJMW_MW',
    index='Month',
    columns='Hour',
    aggfunc='mean'
)

plt.figure(figsize=(15,7))

sns.heatmap(heatmap_data)

plt.xlabel('Hour of Day')
plt.ylabel('Month')
plt.title('Average Power Consumption by Month and Hour')

plt.show()
```

The heatmap shows average power consumption for different combinations of month and hour. Darker and lighter areas help identify periods of higher and lower electricity demand.

```python
hour_daytype = data.groupby(['DayType', 'Hour'])['PJMW_MW'].mean().reset_index()

hour_daytype.head()
```

The data is grouped by hour and day type to compare hourly electricity demand on weekdays and weekends.

```python
plt.figure(figsize=(12,6))

sns.lineplot(
    data=hour_daytype,
    x='Hour',
    y='PJMW_MW',
    hue='DayType'
)

plt.xlabel('Hour of Day')
plt.ylabel('Average Power Consumption (MW)')
plt.title('Hourly Power Consumption: Weekday vs Weekend')

plt.xticks(range(24))

plt.show()
```

This graph compares hourly electricity consumption patterns between weekdays and weekends. It helps determine whether peak demand occurs at different times depending on the type of day.

```python
max_consumption = data.loc[data['PJMW_MW'].idxmax()]

max_consumption
```

This identifies the record with the highest electricity consumption in the dataset. It helps us understand the time and demand level corresponding to the maximum observed consumption.

```python
plt.figure(figsize=(10,5))

sns.histplot(data['PJMW_MW'], bins=30, kde=True)

plt.xlabel('Power Consumption (MW)')
plt.ylabel('Frequency')
plt.title('Distribution of Power Consumption')

plt.show()
```

The histogram shows how frequently different power consumption values occur. It helps us understand the distribution of electricity demand and identify whether most observations are concentrated around a particular range.

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=data['PJMW_MW'])

plt.xlabel('Power Consumption (MW)')
plt.title('Boxplot of Power Consumption')

plt.show()
```

A boxplot is used to identify unusually high or low power consumption values. These observations are called potential outliers.

```python
data['Rolling_24'] = data['PJMW_MW'].rolling(window=24).mean()

data[['Datetime', 'PJMW_MW', 'Rolling_24']].head(30)
```

A 24-hour rolling average is calculated to smooth short-term fluctuations in electricity consumption. Since the data is hourly, a window of 24 represents approximately one day.

```python
plt.figure(figsize=(15,5))

plt.plot(
    data['Datetime'].tail(1000),
    data['PJMW_MW'].tail(1000),
    label='Actual Consumption'
)

plt.plot(
    data['Datetime'].tail(1000),
    data['Rolling_24'].tail(1000),
    label='24 Hour Rolling Average'
)

plt.xlabel('Date')
plt.ylabel('Power Consumption (MW)')
plt.title('Power Consumption and 24-Hour Rolling Average')

plt.legend()

plt.show()
```

The actual hourly consumption and its 24-hour rolling average are plotted together. The rolling average smooths short-term fluctuations and makes the underlying trend easier to observe.

```python
time_difference = data['Datetime'].diff()

time_difference.value_counts().head()
```

The difference between consecutive timestamps is calculated. Since the dataset contains hourly observations, most consecutive records should have a one-hour difference. This check helps identify possible gaps or irregular time intervals.

```python
expected_range = pd.date_range(
    start=data['Datetime'].min(),
    end=data['Datetime'].max(),
    freq='h'
)

missing_dates = expected_range.difference(data['Datetime'])

len(missing_dates)
```

An expected hourly datetime range is created from the first timestamp to the last timestamp. It is then compared with the timestamps in the dataset to identify missing hourly records.

```python
missing_dates[:20]
```

This displays the first 20 missing timestamps identified in the dataset. If missing timestamps are present, they need to be considered during data preprocessing and model building.

```python
data[['PJMW_MW', 'Year', 'Month', 'Day', 'Hour', 'DayOfWeek']].corr()
```

Correlation measures the linear relationship between numerical variables. This provides an initial understanding of how the time-based features are related to power consumption.

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    data[['PJMW_MW', 'Year', 'Month', 'Day', 'Hour', 'DayOfWeek']].corr(),
    annot=True
)

plt.title('Correlation Heatmap')

plt.show()
```

The correlation heatmap provides a visual representation of the relationships between power consumption and the time-based variables.

```python
daily_data = data.set_index('Datetime')['PJMW_MW'].resample('D').mean()

daily_data.head()
```

The hourly electricity consumption is resampled into daily data by calculating the average consumption for each day. This helps us study the overall daily demand trend.

```python
plt.figure(figsize=(15,5))

plt.plot(daily_data)

plt.xlabel('Date')
plt.ylabel('Average Power Consumption (MW)')
plt.title('Daily Average Power Consumption')

plt.show()
```

This graph shows the average electricity consumption for each day over the entire dataset. It makes longer-term patterns and seasonal fluctuations easier to observe.

```python
daily_max = data.set_index('Datetime')['PJMW_MW'].resample('D').max()

daily_max.head()
```

The maximum hourly electricity consumption for each day is calculated. This helps us understand the daily peak demand.

```python
plt.figure(figsize=(15,5))

plt.plot(daily_max)

plt.xlabel('Date')
plt.ylabel('Maximum Power Consumption (MW)')
plt.title('Daily Peak Power Consumption')

plt.show()
```

This graph shows the maximum hourly electricity demand for each day. It is useful for understanding peak electricity requirements.
